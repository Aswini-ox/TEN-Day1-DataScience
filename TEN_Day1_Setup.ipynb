```python
import pandas as pd

data = {
    "Name": ["Aswini", "Charu", "Aishwarya"],
    "Marks": [85, 90, 95],
    "Department": ["CSE", "CSE", "CSE"]
}

df = pd.DataFrame(data)

print("Student Data:")
display(df)

print("Average Marks:", df["Marks"].mean())
```

```python
# Display the first two rows
print("First 2 Students:")
display(df.head(2))

# Find students scoring above 85
print("Students scoring above 85:")
display(df[df["Marks"] > 85])

# Sort students by marks
print("Students sorted by marks:")
display(df.sort_values(by="Marks", ascending=False))
```
